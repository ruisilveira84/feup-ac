
# Evaluation Metric

## Overview
This notebook implements the custom evaluation metric for the playoff prediction task, using the real dataset.

## Metric Description
1. For each season, the model predicts probabilities for each team qualifying for the playoffs.
2. These probabilities are scaled such that the sum equals 8 (number of playoff slots).
3. The adjusted predictions are compared to the true labels (`playoff`), and the error is calculated as the sum of absolute differences.
    

## Metric Calculation

```python

import pandas as pd
import numpy as np

def adjust_predictions(predictions, num_playoff_slots=8):
    """
    Adjusts predictions so that their sum equals the number of playoff slots.
    """
    adjusted_predictions = num_playoff_slots * (predictions / predictions.sum())
    return adjusted_predictions

def evaluation_metric(predictions, true_labels, num_playoff_slots=8):
    """
    Calculates the custom evaluation metric based on adjusted predictions.
    """
    adjusted_predictions = adjust_predictions(predictions, num_playoff_slots)
    error = np.sum(np.abs(adjusted_predictions - true_labels))
    return error
```


## Testing with Data
    

```python

import sys
import os

# Carregar os dados do pipeline (ajustar conforme necessário)
sys.path.append(os.path.abspath('../src'))
from process_data import add_features
import read_data as rd

# Ler e preparar os dados
tables = rd.read_data()
teams = add_features(tables["teams"])

# Extrair probabilidades e rótulos verdadeiros
# Exemplo: vamos supor que `predictions` seja a saída do modelo
# e `playoff` é a coluna que indica classificação (1 = 'Y', 0 = 'N').

# Simular probabilidades do modelo para cada time
predictions = teams["win_rate"]  # Substituir pelo modelo real
true_labels = teams["playoff"].apply(lambda x: 1 if x == "Y" else 0)

# Ajustar predições e calcular erro
adjusted_preds = adjust_predictions(predictions)
error = evaluation_metric(predictions, true_labels)

print("Adjusted Predictions:", adjusted_preds)
print("Evaluation Error:", error)

    
```
